```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

keys = [
    "CE40km_1p0MW_Aplus_coat_flow_10",
    "CE40km_1p0MW_Aplus_coat_flow_5",
    "CE40km_1p0MW_Aplus_coat_flow_7",
    "CE40km_1p0MW_aLIGO_coat_flow_10",
    "CE40km_1p0MW_aLIGO_coat_flow_5",
    "CE40km_1p0MW_aLIGO_coat_flow_7",
    "CE40km_1p5MW_Aplus_coat_flow_10",
    "CE40km_1p5MW_Aplus_coat_flow_5",
    "CE40km_1p5MW_Aplus_coat_flow_7",
    "CE40km_1p5MW_aLIGO_coat_flow_10",
    "CE40km_1p5MW_aLIGO_coat_flow_5",
    "CE40km_1p5MW_aLIGO_coat_flow_7"
]

# 3 rows × 4 columns = 12 plots
fig, axes = plt.subplots(3, 4, figsize=(18, 12))

with h5py.File("bbh_samples.h5", "r") as f:
   
    print(list(f.keys()))
   
    for i, key in enumerate(keys):
        row = i // 4   # 4 columns
        col = i % 4
       
        dataset_name = f"snr_{key}"
       
        snr = np.array(f[dataset_name])
        snr_filtered = snr[snr > 800]
       
        axes[row, col].hist(snr_filtered, bins=40)
        axes[row, col].set_title(key, fontsize=8)
        axes[row, col].set_xlabel("SNR > 800")
        axes[row, col].set_ylabel("Count")

plt.tight_layout()

# SAVE BEFORE SHOW
plt.savefig("snr_histograms_CE40.png", dpi=300, bbox_inches='tight')

plt.show()
```

```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

# Open file
with h5py.File("bbh_samples.h5", "r") as f:
    print(list(f.keys()))
    
    # Load individual SNR arrays
    snr1 = np.array(f['snr_CE40km_1p5MW_Aplus_coat_flow_5'])
    snr2 = np.array(f['snr_CE20km_1p5MW_Aplus_coat_flow_5'])
    snr3 = np.array(f['snr_A1'])

# Combine in quadrature
snr_combined = np.sqrt(snr1**2 + snr2**2 + snr3**2)

# Filter
snr_filtered = snr_combined[snr_combined > 800]

# Count events
count = len(snr_filtered)
print(f"Number of events with combined SNR > 800: {count}")

# Plot
plt.hist(snr_filtered, bins=50)
plt.xlabel("Combined SNR")
plt.ylabel("Count")
plt.title(f"Combined SNR (1.5 MW flow 5)  > 800  (N = {count})")
# SAVE BEFORE SHOW
plt.savefig("snr_1p5MW_flow5.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
import h5py
import numpy as np

with h5py.File("bbh_samples.h5", "r") as f:
    
    # --- Load SNRs ---
    snr1 = np.array(f['snr_CE40km_1p5MW_Aplus_coat_flow_5'])
    snr2 = np.array(f['snr_CE20km_1p5MW_Aplus_coat_flow_5'])
    snr3 = np.array(f['snr_A1'])
    
    # --- Find common length ---
    #N = min(len(snr1), len(snr2), len(snr3), len(f['geocentric_time']))

    N = 50000
    
    # --- Trim everything ---
    snr1 = snr1[:N]
    snr2 = snr2[:N]
    snr3 = snr3[:N]
    
    # --- Combined SNR ---
    snr_combined = np.sqrt(snr1**2 + snr2**2 + snr3**2)
    mask = snr_combined > 800
    
    print(f"Total events considered: {N}")
    print(f"Selected events: {np.sum(mask)}")
    
    # --- Parameters (trimmed) ---
    chirp_mass = np.array(f['detector_frame_chirp_mass'])[:N][mask]
    chirp_mass_sigma = np.full_like(chirp_mass, 1e-2)
    
    mass_ratio = np.array(f['q'])[:N][mask]
    mass_ratio_sigma = np.full_like(mass_ratio, 1e-1)
    
    chi_1 = np.array(f['chi_1'])[:N][mask]
    chi_1_sigma = np.full_like(chi_1, 0.2)
    
    chi_2 = np.array(f['chi_2'])[:N][mask]
    chi_2_sigma = np.full_like(chi_2, 0.2)
    
    ra = np.array(f['ra'])[:N][mask]
    dec = np.array(f['dec'])[:N][mask]
    
    luminosity_distance = np.array(f['luminosity_distance'])[:N][mask]
    luminosity_distance_sigma = np.full_like(luminosity_distance, 15.0)
    
    theta_jn = np.array(f['theta_jn'])[:N][mask]
    psi = np.array(f['psi'])[:N][mask]
    phase = np.array(f['phase'])[:N][mask]
    
    geocent_time = np.array(f['geocentric_time'])[:N][mask]
    geocent_time_sigma = np.full_like(geocent_time, 1e-4)

# --- Save ---
with open("index_1p5MW_flow5.txt", "w") as out:
    for i in range(len(chirp_mass)):
        out.write(
            f"--chirp_mass {chirp_mass[i]}\t"
            f"--chirp_mass_sigma {chirp_mass_sigma[i]}\t"
            f"--mass_ratio {mass_ratio[i]}\t"
            f"--mass_ratio_sigma {mass_ratio_sigma[i]}\t"
            f"--chi_1 {chi_1[i]}\t"
            f"--chi_1_sigma {chi_1_sigma[i]}\t"
            f"--chi_2 {chi_2[i]}\t"
            f"--chi_2_sigma {chi_2_sigma[i]}\t"
            f"--ra {ra[i]}\t"
            f"--dec {dec[i]}\t"
            f"--luminosity_distance {luminosity_distance[i]}\t"
            f"--luminosity_distance_sigma {luminosity_distance_sigma[i]}\t"
            f"--theta_jn {theta_jn[i]}\t"
            f"--psi {psi[i]}\t"
            f"--phase {phase[i]}\t"
            f"--geocent_time {geocent_time[i]}\t"
            f"--geocent_time_sigma {geocent_time_sigma[i]}\t\n"
        )

print("✅ Saved successfully to index_1p5MW_flow5.txt")
```
